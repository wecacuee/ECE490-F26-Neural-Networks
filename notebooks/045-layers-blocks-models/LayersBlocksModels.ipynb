## Modern neural networks are made up of non-linear layers

We saw a linear model in both regression and perceptron $y = m x + c$ which can be extended to 3D as
$$ y_1 = w_1 x_1 + w_2 x_2 + w_0$$ and n-dimensional input with m-dimensional output as

$$ \begin{bmatrix} y_1\\
y_2\\
\vdots\\
y_n
\end{bmatrix} 
= \begin{bmatrix} 
w_{11} & w_{12} & \dots & w_{1m}\\
w_{21} & w_{22} & \dots & w_{2m}\\
w_{n1} & w_{n2} & \dots & w_{nm}
\end{bmatrix} \begin{bmatrix} x_1\\
x_2\\
\vdots\\
x_m
\end{bmatrix} 
+
\begin{bmatrix} b_1\\
b_2\\
\vdots\\
b_n
\end{bmatrix}.
$$

This can be succintly written in vector algebra as $\bfy \in \bbR^n, W \in \bbR^{n \times m}, \bfx \in \bbR^m, \bfb \in \bbR^m$,
$$ \bfy = W \bfx + \bfb.$$

This is a general equation of Linear layer, but it is technically not a linear function, this is an affine function. A Linear layer is also known as a Dense Layer or a Feed Forward layer.

However two linear layers cannot be combined together, because combination of two linear layers is another Linear layer without increase in representation power.

Consider the two layers where the output of one is the input to the other,
\begin{align} \bfy = W_1 \bfx + \bfb_2, \qquad  \bfz = W_2 \bfy + \bfb_1. \end{align}

This is equivalent to a single Linear Layer without any increase in representation power.
\begin{align} \bfz = \underbrace{W_2 W_1}_{\text{another matrix}} \bfx + \underbrace{(W_2 \bfb_1 + \bfb_2)}_\text{another vector}. \end{align}

### Non linear activation functions



To address this problem all layers must be made non-linear before they are composed together using a non-linear activation function. There are many many options for this. A lot of them are listed in [Pytorch Documentation](https://docs.pytorch.org/docs/stable/nn.html#non-linear-activations-weighted-sum-nonlinearity). We will consider four of them,

1. Rectified Linear Unit: $$\text{ReLU}(x) = \begin{cases} x &\text{if }x \ge 0 \\ 0 &\text{otherwise }\end{cases}$$
2. Tangent hyperbolic: $$\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$$
3. Sigmoid: $$\sigma(x) = \frac{1}{1+e^{-x}}$$
4. Sigmoid Linear Unit: $$\text{SiLU}(x) = \frac{x}{1+e^{-x}}$$

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotalt import show_with_alt
def relu(x): return np.maximum(0, x)
def sigmoid(x): return 1/(np.exp(-x)+1)
def silu(x): return x/(np.exp(-x)+1)
activations = dict(relu=relu, tanh=np.tanh, sigmoid=sigmoid, silu=silu)
x = np.linspace(-5, 5, 100)
for (name, func), color in zip(activations.items(), 'rgbk'):
    plt.plot(x, func(x), color, label=name)
plt.grid()
plt.legend()
show_with_alt()
```

Let $a(\bfx)$ be any of the activation function, then any Linear layer becomes a non-Linear layer by composition with a non-Linear activation function,
$$ \bfy = a(W \bfx + \bfb) $$

### Multi-layer Perceptron

A two layer perceptron is given by composing two linear layers with a non-Linear activation layer in the middle:

$$ \hat{\bfy} = W_2 a(W_1\bfx + \bfb_1) + \bfb_2 $$

which is same as generating a hidden-layer and using it to generate the output

$$ \bfh = a(W_1\bfx + \bfb_1)$$
$$ \hat{\bfy} = W_2  \bfh + \bfb_2.$$

The size of the hidden-layer determines the weight matrix dimensions $W_1$, $\bfb_1$, $W_2$, and $\bfb_2$.

An MNIST example that uses two-layer perceptron is provided here: [MLPUsingPytorch](MLPUsingPytorch.html). The two-layer Multi-layer Perceptron (MLP) using Pytorch is provided here:

```python
%%script false --no-raise-error
import torch.nn as tnn

# Inherit all neural networks from tnn.Module
class MLPNetwork(tnn.Module):
    def __init__(self, hidden_size=10, nclasses=10, input_size=28*28):
        super().__init__()
        self._layers = torch.nn.ModuleList([torch.nn.Flatten(),
            tnn.Linear(input_size, hidden_size), # h1 = W1 @ x + b1
            tnn.ReLU(),                          # h2 = ReLU(h1)
            tnn.Linear(hidden_size, nclasses)])  # h3 = W2 @ h2 + b2
        
    def forward(self, x):
        # Initialiize layers in __init__ function, but the layers
        # are executed on the input x in the forward function
        for l in self._layers:
            xnext = l(x) # call the layers in sequence
            x = xnext
        return x
model = MLPNetwork()
```

Alternatively you can use `tnn.Sequential` for a sequence of layers.

```python
%%script false --no-raise-error
# alternatively you can also
hidden_size=10
nclasses=10
input_size=28*28
model = tnn.Sequential(torch.nn.Flatten(),      # converts 28x28 images into876
           tnn.Linear(input_size, hidden_size), # h1 = W1 @ x + b1
           tnn.ReLU(),                          # h2 = ReLU(h1)
           tnn.Linear(hidden_size, nclasses))   # h3 = W @ h2 + b2
```
