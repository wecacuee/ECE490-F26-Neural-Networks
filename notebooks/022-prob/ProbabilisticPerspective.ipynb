# Optimization perspective is not enough

So far in this course, we stuck to an optimization perspective on Machine Learning. We identified the following steps:

1. Collecting a training dataset $\calD = \{ (\bfx_1, \bfy_1), \dots, (\bfx_n, \bfy_n) \}$
2. Selecting a model $\hat{\bfy} = f(\bfx; \bfw)$ with weights (also called the parameters) $\bfw$.
3. Selecting a loss function $l(\bfy_i, \hat{\bfy}_i)$
4. Training: Minimizing the total or average loss function to find the optimal weights

\begin{align}
\bfw^* &= \arg~\min_{\bfw} \frac{1}{n} \sum_{i=1}^n l(\bfy_i, \hat{\bfy}_i; \bfw) \\
&= \arg~\min_{\bfw} \underbrace{\frac{1}{n} \sum_{i=1}^n l(\bfy_i, f(\bfx_i; \bfw))}_{L(\calD; \bfw)}
\end{align}

5. For a new input $\bfx^* \notin \calD$, use $\hat{\bfy}^* = f(\bfx^*; \bfw^*)$ as the prediction.

For linear models (or other "simple" models) this perspective works fine. However, in general it does not.

**Question**

Imagine you are designing a pedestrian detector (input: image, output: a box around all pedestrians in the image) for an autonomous car (or a face detector for your ring-camera based security system). You collect a training dataset of pedestrians in Iceland and then try to deploy it in India. 

- How well is your pedestrian detector likely to work? Very good, very bad, okayish.
- Why?
- How would describe the reason mathematically?

**Answer**

The pedestrian detector is likely to be very bad to okayish.

Because pedestrians in Iceland are likely to look different from pedestrians in India. [Evidence from facial recognition](https://www.scientificamerican.com/article/police-facial-recognition-technology-cant-tell-black-people-apart/).

The real world data is called test data. Mathematically, all the samples in both training data and test data must be identically distributed. In other words, they must have the same probability distribution. Let $\bbP(X, Y)$ be the true but unknown probability distribution. All samples, in the training data and the test data must come from the same distribution, $(\bfx_i, \bfy_i) \sim \bbP(X, Y)$ and $(\bfx^*, \bfy^*) \sim \bbP(X, Y)$. 

From a Machine Learning engineer perspective, you can try to get close to this ideal by carefully collecting your training data so that matches the test data as faithfully as possible.

Since we need probabilistic understanding here, you can review some [Probability basics here](ProbabilityBasics.html)

## Independent and Identically distributed


A random variable $Z_1$ is said to be independent from random variable $Z_2$, denoted as $Z_1 \perp Z_2$ if either of the three equivalent conditions are met:

1. $$\bbP(Z_1, Z_2) = \bbP(Z_1) \bbP(Z_2)$$
2. $$\bbP(Z_1 | Z_2) = \frac{\bbP(Z_1, Z_2)}{\bbP(Z_2)} = \bbP(Z_1)$$
3. $$\bbP(Z_2 | Z_1) = \frac{\bbP(Z_1, Z_2)}{\bbP(Z_1)} = \bbP(Z_2)$$

In typical machine learning, we want the data samples to be Independent and Identically Distributed, in short the i.i.d  assumption.

## Maximum Likelihood estimation

The optimization perspective can be related to maximum likelihood estimation under the iid (Independent and Indentically distributed) assumpotion.

Let the $\bfx_i$ and $y_i$ be random vectors for all $i$. Model the probability distribution as a negative log of the loss function:

$$ P((\bfx_i, y_i)| \bfW) = \frac{1}{Z} \exp(-l(y_i, f(\bfx_i; \bfW)).$$

If the samples are IID, then we can write the probability of the entire dataset as products of sample probabilities

$$ P(\calD|\bfW) = \prod_{i=1}^n P((\bfx_i, y_i)| \bfW) $$

$$ P(\calD|\bfW) = \prod_{i=1}^n \frac{1}{Z} \exp(-l(y_i, f(\bfx_i; \bfW)).$$

A product of exponents is the summation of their powers,

$$ P(\calD|\bfW) = \frac{1}{Z} \exp(-\sum_{i=1}^n l(y_i, f(\bfx_i; \bfW)).$$

Denote $$ L(\calD; \bfW) = \sum_{i=1}^n l(y_i, f(\bfx_i; \bfW).$$

Minimizing the loss is same as maximizing the probability $P(\calD|\bfW)$.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotalt import show_with_alt

if DEBUG: np.random.seed(43)

def lstsqsvd(X, y):
    ndata, nweights = X.shape
    if ndata >= nweights:
        U, s, Vh = np.linalg.svd(X, full_matrices=False, compute_uv=True, hermitian=False)
        rank = s.shape[-1]
        if rank < nweights: print(f"Rank: {rank} <  Weights: {nweights}")
        return Vh.T @ np.diag(1/s) @ U.T @ y
    else:
        U, s, Vh = np.linalg.svd(X, full_matrices=False, compute_uv=True, hermitian=False)
        rank = s.shape[-1]
        if rank < ndata: print(f"Rank: {rank} <  Data: {ndata}")
        return Vh.T @ np.diag(1/s) @ U.T @ y



class LinearRegression:
    def __init__(self, reg=0.1):
        self.w = None
        self.mean = None
        self.std = None
        self.reg = reg
        self.ymean = None
        self.ystd = None
        self.eps = None
        self.target_std = 0.3
        
    def fit(self, X, y):
        self.eps = np.finfo(y.dtype).eps
        
        self.mean = X.mean()
        self.std = X.std()
        Xnormed = (X - self.mean) * self.target_std / self.std
        X_and_1 = np.hstack((Xnormed, np.ones((*X.shape[:-1], 1))))
        
        self.ymean, self.ystd = y.mean(), y.std()
        ynormed = (y - self.ymean) * self.target_std / self.ystd
        self.w, res, rank, s = np.linalg.lstsq(X_and_1, ynormed, rcond=self.eps)
        if DEBUG: 
            print(f"""We want rank to min(ndata, nweights) = min({X_and_1.shape[-2]}, {X_and_1.shape[-1]})."""
                  f""" Check if the singular values s are consistently smaller than {self.eps}""")
            print("Rank: ", rank)
            print("Singular values: ", s)

        y_pred_normed = (X_and_1 @ self.w) 
        
        y_pred = y_pred_normed * self.ystd / self.target_std + self.ymean
        
        res = self.error(y_pred, y)
        return self.w, res

    def predict(self, X):
        Xnormed = (X - self.mean) * self.target_std / self.std
        y_pred_normed = (Xnormed @ self.w[:-1] + self.w[-1])
        return y_pred_normed * self.ystd / self.target_std + self.ymean
        
    def error(self, y_pred, y_true):
        return (np.abs(y_true - y_pred).mean()) / np.abs(y_true).mean()
        

class PolynomialFeatures:
    def __init__(self, degree):
        self.degree = degree
        self.features = None
        self.maxdegree = None

    def get_features(self, X):
        # For x < 1, we want x**max_degree > eps to avoid unnecessary low ranks
        self.maxdegree = int(round(-np.log10(np.finfo(X.dtype).eps)))-1 # 15
        return np.concatenate([ np.pow(X, i)
                for i in np.linspace(1, self.maxdegree, self.degree)],
                                       axis=-1)

class FourierFeatures:
    def __init__(self, degree):
        self.degree = degree
        self.features = None

    def get_features(self, X):
        return np.concatenate([ np.cos(2*np.pi*i*X/self.degree)
                for i in np.arange(self.degree)] + 
                              [ np.sin(2*np.pi*i*X/self.degree)
                for i in np.arange(self.degree)],
                                       axis=-1)

class Pipeline:
    def __init__(self, *features_regressor):
        *feature_extractors, regressor = features_regressor
        self.feature_extractors = feature_extractors
        self.regressor = regressor

    def fit(self, X, y):
        for feat_extract in self.feature_extractors:
            X = feat_extract.get_features(X)
        return self.regressor.fit(X, y)

    def predict(self, X):
        for feat_extract in self.feature_extractors:
            X = feat_extract.get_features(X)
        return self.regressor.predict(X)
        
    def error(self, y_pred, y_true):
        return self.regressor.error(y_pred, y_true)

def make_pipeline(*args):
    return Pipeline(*args)

def truefun(X):
    return (3+X+ 2*np.sin(2*X))

    
X = np.linspace(0, 10, 30).reshape(-1, 1)
y = truefun(X).ravel() + np.random.normal(0, 1.2, X.shape[0])

X_plot = np.linspace(0, 10, 100).reshape(-1, 1)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

degrees = [5, 11, 35]
titles = [f'Underfitting\n(High Bias: {degrees[0]})', f'Right Fit\n(Balanced: {degrees[1]})', f'Overfitting\n(High Variance: {degrees[2]})']

for ax, degree, title in zip(axes, degrees, titles):
    model = make_pipeline(FourierFeatures(degree), LinearRegression())
    _, train_err = model.fit(X, y)
    train_err = train_err.mean()
    y_pred = model.predict(X_plot)
    y_true = truefun(X_plot).ravel()
    eval_err = model.error(y_pred, y_true).mean()
    
    ax.scatter(X, y, color='blue', alpha=0.6, label='Data points')
    ax.plot(X_plot, y_pred, color='red', linewidth=2, label='Model')
    ax.plot(X_plot, truefun(X_plot), color='green', linestyle='--', alpha=0.7, label='True function')
    ax.set_title(title, fontsize=12)
    ax.text(0, 15, f'ee: {eval_err:.3f}, te: {train_err:.3f}')
    ax.set_xlabel('X')
    ax.set_ylabel('y')
    ax.legend(loc='upper right', fontsize=8)
    ax.set_ylim(-.5, 15.5)

plt.tight_layout()
#plt.savefig('overfitting_visualization.png', dpi=150)
show_with_alt(fig=fig, alt_text="""Under fitting happens when a curve does not
make use of all the data; overfitting happens when the curve fits even 
the noise of the data instead of ignoring it. A good curve fit happens when the
curve ignores the noise but fits most of the desired pattern.""")
```

### Ways to avoid overfitting

1. Use an additional part in the loss function called  a regularizer. For example in a linear least squares problem you can prefer the "simplest" solution by  adding  another term $\|\lambda \bfw\|^2$,
    $$ \bfw^* = \arg~\min_\bfw \|\bfy - \bfX \bfw\|^2 + \lambda \|\bfw\|^2, $$
   
#### Exercise

Equate the gradient of the loss to zero in order to find the optimal weight.
    $$ \bfw^* = \arg~\min_\bfw \|\bfy - \bfX \bfw\|^2 + \lambda \|\bfw\|^2, $$

#### Solution

$$  = \bfy^\top \bfy + 2\bfy^\top \bfX \bfw + \bfw^\top (\bfX^\top \bfX + \lambda \bfI) \bfw$$
$$ \implies \bfw^* = (\bfX^\top \bfX + \lambda \bfI)^{-1} \bfX^\top \bfy$$ 

#### Exercise

Equate the gradient of the loss to zero in order to find the optimal weight.

$$ \bfw^* = \arg~\min_\bfw {1 \over 2} \|\bfw\|^2 \text{ s.t. } \bfy = \bfX \bfw, $$
$$ \bfw^* = \arg~\min_\bfw {1 \over 2} \|\bfw\|^2 + \Lambda^\top (\bfy - \bfX \bfw) $$

#### Solution

$$ \bfw - \bfX^\top \Lambda =  0,\qquad  \bfy - \bfX \bfw = 0$$
$$ \bfy - \bfX \bfX^\top \Lambda = 0$$
$$ \Lambda = (\bfX \bfX^\top)^{-1} \bfy $$
$$ \bfw = \bfX^\top(\bfX \bfX^\top)^{-1} \bfy $$

## Expectation vs Mean

Given i.i.d. random variables $Z_1, Z_2, ... Z_n$ with distribution $\bbP(.)$, write the mathematical definition for the mean and that of the expectation. What is the difference?

**Question**

What is the diference between Empirical risk and Expected risk?

**Answer**

Empirical risk is the average loss over the dataset while Expected risk is the expectation of the loss when the inputs and outputs are treated as random variables. 

For example, if a dataset $\calD = \{(\bfx_1, y_1), \dots, (\bfx_n, y_n)\}$ with $\bfx_i \in \bbR^n$ and $y_i \in \bbR$ is given and the model $f(\bfx_i; \bfW)$ predicts labels $\hat{h}_i = f(\bfx_i; \bfW)$. Then the empirical risk is the average loss over the dataset samples,

$$ R_\calD(\bfW) = \frac{1}{n} \sum_{i=1}^n loss(\hat{y}_i, y_i). $$

On the other hand to compute Expected risk, we assume that the dataset $\calD$ was independently sampled (IID assumption) from a distribution $\bbP_{X,Y}$ where the input $\bfx_i$ is an instance of random variable $X$ and $y_i$ is an instance of random variable $Y$.

$$ R(\bbP_{X,Y}, \bfW) = \bbE_{X,Y} [ loss(\hat{Y}, Y) ], $$
where $\hat{Y} = f(X; \bfW)$.

If $X$ and $Y$ are continous random variables, then the expectation is the weighted integraal over all possible values of values of $X$ and $Y$ weighted by the joint probability density function (PDF) of $X$ and $Y$.

$$ R(\bbP_{X,Y}, \bfW) = \bbE_{X,Y} [ loss(\hat{Y}, Y) ] = \int_{x \in \Omega_X} \int_{y \in \Omega_Y} loss(\hat{y}, y) f_{X,Y}(x, y) dx dy, $$
where $\hat{y} = f(x; \bfW)$ and $\Omega_X$ and $\Omega_Y$ is the sample space for $X$ and $Y$.

## Generalization gap bounds in Machine Learning

There are many generalization gap bounds in machine learning, often studied under Machine Learning theory; Probably Approximatly Correct (PAC) theory. Mohri 2012, Foundations of Machine Learning is an excellent book to learn more about it.


If the loss $l(.,.)$ is between [0, 1], then for any $\delta > 0$ with Probability at least $1-\delta$, we have

\begin{align}
\max_{\bfW \in \calW} R(\bbP_{X,Y}, \bfW) - R_\calD(\bfW) \le 2 \mathfrak{R}_m(l) + \sqrt{\frac{-\ln(\delta)}{2n}}
\end{align}

where Rademacher Complexity is defined as:
$$ \mathfrak{R}_m(G) = \bbE_{S \sim D^m}\left[ \hat{\mathfrak{R}}_S(G) \right] = \bbE_{S \sim D^m}\left[ \bbE_{\bfsigma}\left[ \sup_{g \in G} \frac{1}{m} \sum_{i=1}^m \sigma_i g(z_i) \right]  \right] $$

**Definition 3.1 Empirical Rademacher complexity**

Let $G$ be a family of functions mapping from $Z$ to $[a, b]$ and $S = (z_1, \dots, z_m)$ a fixed samples of size $m$ with elements in $Z$. Then, the empirical Rademacher complexity of $G$ with respect to the sample $S$ is defined as:

$$ \hat{\mathfrak{R}}_S(G) = \bbE_{\bfsigma}\left[ \sup_{g \in G} \frac{1}{m} \sum_{i=1}^m \sigma_i g(z_i) \right] $$

where $\bfsigma = (\sigma_1, \dots, \sigma_m)^\top$, with $\sigma_i$s independent uniform random variables taking values in $\{-1, +1\}$. The random variables $\sigma_i$ are called Rademacher variables.

**Definition 3.2 Rademacher complexity**

Let $D$ denote the distribution according to which samples are drawn. For any integre $m \ge 1$, the Rademacher complexity of $G$ is the expectation of the empirical Rademacher complexity over all samples of size $m$ drawn according to $D$:
$$ \mathfrak{R}_m(G) = \bbE_{S \sim D^m}\left[ \hat{\mathfrak{R}}_S(G) \right]. $$

## Training, validation and test data

Since we can never compute expectation over loss that we actually want to compute:
$ R(\bbP_{X,Y}, \bfW) = \bbE_{X,Y} [ loss(\hat{Y}, Y) ]$, we settle for the next best thing.

In typical machine learning, we divide our given dataset into three parts:

1. Training data $\calD_t$: To actually train the model by gradient descent or other algorithms.
2. Validation data $\calD_v$: To figure out hyper-parameters of the training pipeline, to test for **overfitting**, when to stop training.
3. Test/evaluation data $\calD_e$: Only used for evaluating and comparing algorithms. This test data is sacred. Training/validation data must not be contain any information from test data. If it does, then training data is said to be **contaminated**. This is a huge problem in Large Language Models as big companies compete to get better performance at benchmarks they often end up crawling the answers from the internet.

If we uniformly sample the training, validation and test data from the collected dataset $\calD$ then they are likely to be identically distributed (have the same probability distribution). 
Since, we are training on the training dataset and evaluating on an identically distributed but different test dataset, we can expect our model's real-world performance to match the performance on the test dataset.

## From Gradient descent to Stochastic Gradient Descent

### Gradient descent method.

**given** a starting point $\bfW \in \text{dom}{f}$ .

**repeat**
1. Choose step size $\alpha_t$ 
2. Update:
   $$\bfW \leftarrow \bfW  - \alpha_t \nabla_\bfW \underbrace{ \left( \frac{1}{|\calD_t|} \sum_{(\bfx_i, \bfy_i) \in \calD_t} l(\bfy_i, f(\bfx_i; \bfW)) \right) }_{\calL(\calD_t; \bfW)}, $$
   or equivalently
   $$\bfW \leftarrow \bfW  -  \frac{\alpha_t }{|\calD_t|} \sum_{(\bfx_i, \bfy_i) \in \calD_t} \nabla_\bfW  l(\bfy_i, f(\bfx_i; \bfW)) $$

**until** stopping criterion is satisfied.

#### Intuition behind Stochastic gradient descent
Note that one update on Gradient descent method can only happen after we have looped over the entire dataset. For large datasets that can take a long time.

Moreover, with the same intuition we have developed using training-validation-test split, a uniformly sampled smaller dataset from the full data gives us somewhat similar information as the full training dataset. With these two ideas, we develop an algorithm called Batch Stochastic Gradient Descent (Batch-SGD).

### Batch Stochastic gradient descent (Batch-SGD) method.

**given** a starting point $\bfW \in \text{dom}{f}$ .

**repeat**
1. Choose step size $\alpha_t$ 
2. **for** e **in** range(n_epochs) do:

    a. Shuffle the dataset.
   
    b. Sample a small batch $\calB_j \subset \calD_t$ from the training dataset $\calD_t$. For each batch do:
       $$\bfW \leftarrow \bfW  -   \alpha_t\nabla_\bfW \underbrace{\left(\frac{1 }{|\calB_j|} \sum_{(\bfx_i, \bfy_i) \in \calB_j}l(\bfy_i, f(\bfx_i; \bfW)) \right)}_{\calL(\calD_t; \bfW)} $$
       or equivalently
       $$\bfW \leftarrow \bfW  -  \frac{\alpha_t }{|\calB_j|} \sum_{(\bfx_i, \bfy_i) \in \calB_j} \nabla_\bfW l(\bfy_i, f(\bfx_i; \bfW)) $$
   
    c. Evaluate the validation loss $\frac{1}{|\calD_v|}\sum_{(\bfx_i, \bfy_i) \in \calD_v} \nabla_\bfw l(\bfy_i, f(\bfx_i; \bfW))$ on validation set $\calD_v$. Decide to stop when the validation loss stops decreasing.


If the size of small batch $|\calB_j|=1$ is just one sample, then the algorithm is called Stochastic Batch gradient descent (SGD).

### Advantages of Batch SGD over Gradient descent (GD)

1. **Faster** on big datasets: SGD makes progress faster without having to wait to compute gradient over the entire dataset. If the dataset is big, SGD makes faster progress.
   
2. **Aligns better** with machine learning objective: In ML the true objective to be minimized is not the *empirical training loss*, instead it is the *expected loss*, hence Batch-SGD avoids overfitting to the training set by adding random choices of small training sets (the batches).

#### When to use Gradient descent

1. Use in optimization, not in machine learning.
2. Use with small datasets and with simpler models when you know the model you want to fit. For example, polynomial regression etc. 

### Bayes Rule

**Question:**

Given two random variables $X$ and $Y$ specify the relationship between $P(X|Y)$ and $P(Y|X)$ using Bayes rule?

**Answer**

$$ P(Y|X) = \frac{P(X|Y)P(Y)}{P(X)}$$

##### Question:
How can regularization can be interpretted as an application of the Bayes theorem?

Given the dataset $\calD = \{ (\bfx_1, y_1), \dots, (\bfx_n, y_n) \}$, a model $\hat{y}_i = f(\bfx_i; \bfW)$, a regularizer $R(\bfW)$ and a loss function $l(y_i, \hat{y}_i)$, show that the following optimization problem can be interpreted as maximum-a-posteriori estimation. In the process show that for the interpretation, we need the IID (independently, identically distributed) assumption over the dataset. List any other assumptions that you need for the interpretation.

$$ \bfW^* = \arg~\min_\bfW \sum_{i=1}^n l(y_i, f(\bfx_i; \bfW)) + \lambda R(\bfW),$$

where $\lambda$ is some positive constant that balances between the loss function and the regularizer.

##### Answer:
Let the $\bfx_i$ and $y_i$ be random vectors for all $i$. Model the probability distribution as a negative log of the loss function:

$$ P((\bfx_i, y_i)| \bfW) = \frac{1}{Z} \exp(-l(y_i, f(\bfx_i; \bfW)).$$

If the samples are IID, then we can write the probability of the entire dataset as products of sample probabilities

$$ P(\calD|\bfW) = \prod_{i=1}^n P((\bfx_i, y_i)| \bfW) $$

$$ P(\calD|\bfW) = \prod_{i=1}^n \frac{1}{Z} \exp(-l(y_i, f(\bfx_i; \bfW)).$$

A product of exponents is the summation of their powers,

$$ P(\calD|\bfW) = \frac{1}{Z} \exp(-\sum_{i=1}^n l(y_i, f(\bfx_i; \bfW)).$$

Denote $$ L(\calD; \bfW) = \sum_{i=1}^n l(y_i, f(\bfx_i; \bfW).$$

The original optimization problem can be written as:
$$ \bfW^* = \arg~\min_\bfW L(\calD; \bfW) + \lambda R(\bfW)$$

Taking negative exponent on both sides turns the problem into a maximization problem because $\exp(-y)$ is a monotonically decreasing function.
$$ \bfW^* = \arg~\max_\bfW \exp(-L(\calD; \bfW))\exp(-\lambda R(\bfW))$$

The first term is the same as maximizing the likelihood $P(\calD|\bfW)$. If we interpret the second term as a prior:
$$ P(\bfW) = \frac{1}{Z'} \exp(-\lambda R(\bfW)),$$

then we can rewrite the original optimization problem as

$$ \bfW^* = \arg~\max_\bfW P(\calD|\bfW) P(\bfW) $$

By Bayes theorem $P(\calD|\bfW) P(\bfW) = P(\bfW|\calD)P(\calD)$, hence we can write the optimization problem as maximizing the posterior

$$ \bfW^* = \arg~\max_\bfW P(\bfW|\calD) P(\calD).$$

We can ignore the evidence term $P(\calD)$, because it is independent of $\bfW$ the optimization variable. The original problem reduces to maximizing the posterior, hence maximum a posteriori:

$$ \bfW^* = \arg~\max_\bfW P(\bfW|\calD)$$

##### Question:
What is a ROC (Receiver Operating characteristics) for a binary classifier? Argue why it lies above the diagonal for any reasonable classifier? Also, argue why it must be concave?

##### Answer:


ROC curve is plot between False positive rate (FPR) $P(\hat{Y}(X) = 1| Y=0)$ and True positive rate (TPR) $P(\hat{Y}(X) = 1| Y = 1)$. 

ROC curve is always *lies above the diagonal* because you can construct a trivial classifier that predicts the positive class with probability $\alpha \in (0, 1)$, i.e. $P(\hat{Y} = 1) = \alpha$ without even looking at the input data $X$. This trivial classifier will form the diagonal of the ROC curve with both FPR and TPR being equal to $\alpha$. Any classifier that works better than this trivial classifier will lie above the diagonal.

*Why ROC curve is a concave curve*

Suppose you are given hyperparameter values on the ROC curve of a classifier such that:

1. The first classifier $1$ has true positive rate (TPR) is $P(\hat{Y}_1(X) = 1| Y = 1) = TPR(f_1)$ at given FPR $P(\hat{Y}_1(X) = 1| Y=0) = f_1$ .

2. Similarly second classifier $2$ has $FPR = P(\hat{Y}_2(X) = 1| Y = 0) = f_2$, assume it's TPR is $TPR(f_2)$.
   
If you pick the classifier $f_1$ randomly with probability $P(\hat{Y} = \hat{Y}_1) = \alpha$ and the other one with probability $P(\hat{Y} = \hat{Y}_2) = 1-\alpha$, and take the output of the picked classifier.

\begin{align} 
P(\hat{Y}(X) = 1 | Y = 1) &= P(\hat{Y}(X) = 1 | \hat{Y} = \hat{Y}_1, Y = 1)P(\hat{Y} = \hat{Y}_1) + P(\hat{Y}(X) = 1 | \hat{Y} = \hat{Y}_2, Y = 1)P(\hat{Y} = \hat{Y}_2) \\
&= P(\hat{Y}(X) = 1 | \hat{Y} = \hat{Y}_1, Y = 1)\alpha + P(\hat{Y}(X) = 1 | \hat{Y} = \hat{Y}_2, Y = 1)(1-\alpha)\\
&= P(\hat{Y}_1(X) = 1 | Y = 1)\alpha + P(\hat{Y}_2(X) = 1 |  Y = 1)(1-\alpha)
&= TPR(f_1) \alpha + TPR(f_2)(1-\alpha)
\end{align}

Then you have FPR of this new classifier as $FPR =P(\hat{Y}_1(X) = 1| Y = 0)\alpha + P(\hat{Y}_2(X) = 1| Y = 0)(1-\alpha) = f_1\alpha + f_2(1-\alpha) $. If you get a better TPR than this randomized classifier, the ROC curve will be concave $TPR^*(\alpha  f_1 + (1-\alpha) f_2) > \alpha TPR(f_1) + (1-\alpha)  TPR(f_2)$. 

The definition of concavity: A function $g$ is concave if for all points $y_1$, $y_2$ and all $\alpha \in (0, 1)$ the following condition is true, $g(y_1 \alpha + y_2 (1-\alpha) > g(y_1)\alpha + g(y_2)(1-\alpha) $.

##### Question:

What is the likelihood ratio test for a binary classifier? Is it better or worse than Posterior ratio test?

##### Answer:
For a binary classifier, likelihood is the probability of observing the evidence $\bfx$ given a hypothesis class $y \in \{0, 1\}$,
$$\text{Likelihood}(\bfx, y) = \bbP(\bfx | y) $$
Likelihood ratio test gives us a classifier that predicts the positive class $y = 1$ if the likelihood ratio exceeds a given threshold $\eta$,

$$
\hat{y}(\bfx) = \begin{cases}
1 & \text{ if } \frac{\bbP(\bfx | y = 1)}{\bbP(\bfx | y = 0)} \ge \eta \\
0 & \text{ otherwise }
\end{cases}
$$

Posterior is the probability of hypothesis class $y$ given an observed evidence $\bfx$,

$$\text{Posterior}(\bfx, y) = \bbP(y | \bfx) = \frac{\bbP(\bfx | y) \bbP(y)}{\bbP(\bfx)}$$

Let the posterior ratio test be,
$$
\hat{y}(\bfx) = \begin{cases}
1 & \text{ if } \frac{\bbP(y = 1 | \bfx)}{\bbP(y = 0| \bfx)} \ge \eta_p \\
0 & \text{ otherwise }
\end{cases}
$$

Posterior ratio test and likelihood ratio test are equivalent, because the condition for posterior ratio test can be written as likelihood ratio test,

\begin{align}
\frac{\bbP(y = 1 | \bfx)}{\bbP(y = 0| \bfx)} &\ge \eta_p\\
\implies
\frac{\bbP(\bfx | y = 1)\bbP(y=1)}{\bbP( \bfx|y=0)\bbP(y=0)} &\ge \eta_p \\
\implies
\frac{\bbP(\bfx | y = 1)}{\bbP( \bfx|y=0))} &\ge \frac{\bbP(y=0)}{\bbP(y=1)}\eta_p.
\end{align}

The last condition is the same condition in the likelihood ratio test with $\eta = \frac{\bbP(y=0)}{\bbP(y=1)}\eta_p$.
