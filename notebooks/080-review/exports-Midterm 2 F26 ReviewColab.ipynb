## Part A: Probabilistic perspective

### The steps of the *optimization perspective* of machine learning.

**Answer**

1. Collect a training dataset $\calD = \{ (\bfx_1, \bfy_1), \dots, (\bfx_n, \bfy_n) \}$.
2. Select a model $\hat{\bfy} = f(\bfx; \bfw)$ with weights (parameters) $\bfw$.
3. Select a loss function $l(\bfy_i, \hat{\bfy}_i)$.
4. Train: minimize the average loss to find the optimal weights,
$$\bfw^* = \arg~\min_{\bfw} \underbrace{\frac{1}{n} \sum_{i=1}^n l(\bfy_i, f(\bfx_i; \bfw))}_{L(\calD; \bfw)}.$$
5. For a new input $\bfx^* \notin \calD$, predict $\hat{\bfy}^* = f(\bfx^*; \bfw^*)$.

For linear and other "simple" models this works fine. In general it is not enough: it says nothing about how the training data relates to the data the model will see after deployment.

### Q3. When is a random variable $Z_1$ independent of a random variable $Z_2$ ($Z_1 \perp Z_2$)? Give three equivalent conditions. What is the i.i.d. assumption?

**Answer**

$Z_1 \perp Z_2$ if any one of these equivalent conditions holds:

1. $\bbP(Z_1, Z_2) = \bbP(Z_1) \bbP(Z_2)$
2. $\bbP(Z_1 | Z_2) = \frac{\bbP(Z_1, Z_2)}{\bbP(Z_2)} = \bbP(Z_1)$
3. $\bbP(Z_2 | Z_1) = \frac{\bbP(Z_1, Z_2)}{\bbP(Z_1)} = \bbP(Z_2)$

The i.i.d. (independent and identically distributed) assumption says the data samples are mutually independent and are all drawn from the same distribution $\bbP(X, Y)$.

### Q4. Given the dataset
$\calD = \{ (\bfx_1, y_1), \dots, (\bfx_n, y_n) \}$, a model $\hat{y}_i = f(\bfx_i; \theta)$, and a loss function $l(y_i, \hat{y}_i)$, show that the following optimization problem can be interpreted as **maximum likelihood estimation**. In the process, show where the i.i.d. assumption is needed. List any other assumptions the interpretation needs.

$$ \theta^* = \arg~\min_\theta \sum_{i=1}^n l(y_i, f(\bfx_i; \theta))$$

**Answer**

Treat $\bfx_i$ and $y_i$ as random vectors. **Assume** the probability of a sample is the exponential of the negative loss:

$$ P((\bfx_i, y_i)| \theta) = \frac{1}{Z} \exp\left(-l(y_i, f(\bfx_i; \theta))\right),$$

where the normalizer $Z$ does not depend on $\theta$ (also an assumption).

**If the samples are i.i.d.**, the probability of the whole dataset is the product of the sample probabilities:

$$ P(\calD|\theta) = \prod_{i=1}^n P((\bfx_i, y_i)| \theta) = \prod_{i=1}^n \frac{1}{Z} \exp\left(-l(y_i, f(\bfx_i; \theta))\right).$$

A product of exponentials is the exponential of the sum of their powers:

$$ P(\calD|\theta) = \frac{1}{Z^n} \exp\left(-\sum_{i=1}^n l(y_i, f(\bfx_i; \theta))\right) = \frac{1}{Z^n} \exp\left(-L(\calD; \theta)\right),$$

where $L(\calD; \theta) = \sum_{i=1}^n l(y_i, f(\bfx_i; \theta))$.

The original problem is $\theta^* = \arg~\min_\theta L(\calD; \theta)$. Because $\exp(-y)$ is monotonically decreasing, this equals

$$ \theta^* = \arg~\max_\theta \exp(-L(\calD; \theta)) = \arg~\max_\theta P(\calD|\theta),$$

which is the maximum likelihood estimate.

### Q5. Given the dataset

$\calD = \{ (\bfx_1, y_1), \dots, (\bfx_n, y_n) \}$, a model $\hat{y}_i = f(\bfx_i; \theta)$, a regularizer $R(\theta)$ and a loss function $l(y_i, \hat{y}_i)$, show that the following optimization problem can be interpreted as **maximum-a-posteriori (MAP) estimation**. Show where the i.i.d. assumption is needed and list any other assumptions.

$$ \theta^* = \arg~\min_\theta \sum_{i=1}^n l(y_i, f(\bfx_i; \theta)) + \lambda R(\theta),$$

where $\lambda > 0$ balances the loss and the regularizer.

*How can regularization be interpreted as an application of Bayes' theorem?*

**Answer**

As in the previous question, assume $P((\bfx_i, y_i)| \theta) = \frac{1}{Z} \exp(-l(y_i, f(\bfx_i; \theta)))$ and that the samples are **i.i.d.**, so that

$$ P(\calD|\theta) = \frac{1}{Z^n} \exp\left(-L(\calD; \theta)\right), \qquad L(\calD; \theta) = \sum_{i=1}^n l(y_i, f(\bfx_i; \theta)).$$

The problem is $\theta^* = \arg~\min_\theta L(\calD; \theta) + \lambda R(\theta)$. Taking $\exp(-\cdot)$, which is monotonically decreasing, turns it into a maximization:

$$ \theta^* = \arg~\max_\theta \exp(-L(\calD; \theta))\exp(-\lambda R(\theta)).$$

The first factor is proportional to the likelihood $P(\calD|\theta)$. **Assume** the second factor is the prior:

$$ P(\theta) = \frac{1}{Z'} \exp(-\lambda R(\theta)).$$

Then

$$ \theta^* = \arg~\max_\theta P(\calD|\theta) P(\theta).$$

By Bayes' theorem, $P(\calD|\theta) P(\theta) = P(\theta|\calD)P(\calD)$, so

$$ \theta^* = \arg~\max_\theta P(\theta|\calD) P(\calD).$$

The evidence $P(\calD)$ does not depend on $\theta$, so it can be dropped:

$$ \theta^* = \arg~\max_\theta P(\theta|\calD),$$

which is maximum a posteriori estimation. The regularizer is the negative log of the prior on the weights. For example, $R(\theta) = \|\theta\|^2$ corresponds to a Gaussian prior.

### Q6. 
Given two random variables $X$ and $Y$, specify the relationship between $P(X|Y)$ and $P(Y|X)$ using Bayes\' rule.

**Answer**

$$ P(Y|X) = \frac{P(X|Y)P(Y)}{P(X)}$$

### Q7. 

You want to find the weights $\bfW$ of a neural network trained on the CIFAR-10 dataset $\calD$. Define the prior, the likelihood and the posterior in terms of $\bfW$ and $\calD$. Write Bayes' rule to get the posterior from the likelihood and the prior.

**Answer**

- **Prior** $P(\bfW)$: the probability of the weights $\bfW$ before seeing the dataset.
- **Likelihood** $P(\calD|\bfW)$: the probability of observing the dataset for a particular choice of weights $\bfW$.
- **Posterior** $P(\bfW|\calD)$: the probability of the weights given the dataset $\calD$.

By Bayes' rule:

$$ P(\bfW|\calD) = \frac{P(\calD|\bfW)P(\bfW)}{P(\calD)}$$

### Q8. 

What does the *No Free Lunch theorem* (Wolpert, 1996) say? What is the problem of *under-determination*?

**Answer**

**Under-determination:** a finite sample of data can be described by quite different models. For example, a straight line and an oscillating curve can both pass through the same points.

**No Free Lunch:** the data collected so far says nothing about the values of the function in regions you have not observed. Averaged over all possible target functions, every learning algorithm has the same expected performance off the training set. An algorithm can only beat others on a restricted subclass of functions (for example, convex functions) whose structure it exploits. In other words, every learner's success depends on its assumptions (inductive bias) matching the problem.

### Q9. 

What are *underfitting* and *overfitting*? How do you tell them apart using training error and evaluation error?

**Answer**

- **Underfitting** (high bias): the model is too simple to capture the pattern in the data. Both training error and evaluation error are high.
- **Overfitting** (high variance): the model fits even the noise in the training data instead of ignoring it. Training error is very low but evaluation error is high.
- **A good fit** ignores the noise but captures most of the desired pattern. Evaluation error is low.

In the polynomial-regression example from class (15 noisy points from $3 + x + 2\sin(9x)$), degree 4 underfits, degree 12 fits well, and degree 30 overfits.

![Underfitting, right fit and overfitting](../022-prob/overfitting_visualization.png)

### Ways to avoid overfitting.

**Answer**

1. Add a **regularizer** to the loss to prefer "simpler" solutions, e.g. for least squares
$$ \bfw^* = \arg~\min_\bfw \|\bfy - \bfX \bfw\|^2 + \lambda \|\bfw\|^2. $$
2. Use a **validation set** to choose hyper-parameters (e.g. model size, $\lambda$) and to decide **when to stop training** (early stopping, when validation loss stops decreasing).
3. Train with **(batch) SGD**: the random choice of small batches keeps the optimizer from fitting the training set exactly.
4. Collect more (representative) training data, or use a smaller model.

### Q11. Find the minimum point of the regularized least-squares (ridge regression) problem

$$\bfw^* = \arg~\min_\bfw \|\bfy - \bfX \bfw\|^2 + \lambda \|\bfw\|^2, $$
where $\bfw \in \bbR^n$, $\bfy \in \bbR^m$, $\bfX \in \bbR^{m \times n}$ and $\lambda \in \bbR^+$.

**Answer**

Let $f(\bfw) = \|\bfy - \bfX \bfw\|^2 + \lambda \|\bfw\|^2$. Write it with inner products:
$$f(\bfw) = (\bfy - \bfX \bfw)^\top(\bfy - \bfX \bfw) + \lambda \bfw^\top \bfw.$$

Expand and collect terms:
$$f(\bfw) = \bfw^\top (\bfX^\top \bfX + \lambda \bfI_n) \bfw - 2\bfy^\top \bfX \bfw + \bfy^\top \bfy.$$

Take the derivative (row-vector convention):
$$\frac{\p}{\p \bfw} f(\bfw) = 2 \bfw^\top(\bfX^\top \bfX + \lambda \bfI_n)  - 2\bfy^\top \bfX.$$

At the minimum the derivative is zero:
$$2 \bfw^{*\top}(\bfX^\top \bfX + \lambda \bfI_n)  - 2\bfy^\top \bfX = \mathbf{0}^\top_n.$$

Transpose and solve:
$$\bfw^* = (\bfX^\top \bfX + \lambda \bfI_n)^{-1} \bfX^\top \bfy.$$

$\bfX^\top \bfX + \lambda \bfI_n$ is always invertible for $\lambda > 0$.

### Q12. 

Find the minimum-norm solution of an under-determined linear system ($\bfX \in \bbR^{m \times n}$, $m < n$, full row rank):

$$ \bfw^* = \arg~\min_\bfw {1 \over 2} \|\bfw\|^2 \text{ s.t. } \bfy = \bfX \bfw. $$

Use the Lagrangian $ {1 \over 2} \|\bfw\|^2 + \Lambda^\top (\bfy - \bfX \bfw) $.

**Answer**

Set the derivatives of the Lagrangian with respect to $\bfw$ and $\Lambda$ to zero:
$$ \bfw - \bfX^\top \Lambda =  0,\qquad  \bfy - \bfX \bfw = 0.$$

Substitute the first into the second:
$$ \bfy - \bfX \bfX^\top \Lambda = 0 \implies \Lambda = (\bfX \bfX^\top)^{-1} \bfy. $$

Therefore
$$ \bfw^* = \bfX^\top(\bfX \bfX^\top)^{-1} \bfy. $$

### Q14. 

Given i.i.d. random variables $Z_1, Z_2, \dots, Z_n$ with distribution $\bbP(\cdot)$, write the mathematical definition of the mean and of the expectation. What is the difference?

**Answer**

The (sample) **mean** is an average over the observed samples:
$$ \bar{Z} = \frac{1}{n} \sum_{i=1}^n Z_i .$$

The **expectation** is a probability-weighted average over all possible values of the random variable:
$$ \bbE[Z] = \sum_{z \in \Omega} z\, \bbP(Z = z) \quad \text{(discrete)}, \qquad \bbE[Z] = \int_{\Omega} z\, f_Z(z)\, dz \quad \text{(continuous)}.$$

The mean is computed from data and is itself a random variable. The expectation is a fixed property of the distribution, which is usually unknown. Under the i.i.d. assumption, the mean converges to the expectation as $n \to \infty$ (law of large numbers).

### Q15. 

What is the difference between *empirical risk* and *expected risk*?

**Answer**

**Empirical risk** is the average loss over the dataset. For a dataset $\calD = \{(\bfx_1, y_1), \dots, (\bfx_n, y_n)\}$ and predictions $\hat{y}_i = f(\bfx_i; \bfW)$:
$$ R_\calD(\bfW) = \frac{1}{n} \sum_{i=1}^n loss(\hat{y}_i, y_i). $$

**Expected risk** is the expectation of the loss when inputs and outputs are random variables. Assume $\calD$ was sampled i.i.d. from $\bbP_{X,Y}$, where $\bfx_i$ is an instance of $X$ and $y_i$ an instance of $Y$:
$$ R(\bbP_{X,Y}, \bfW) = \bbE_{X,Y} [ loss(\hat{Y}, Y) ], \qquad \hat{Y} = f(X; \bfW).$$

For continuous $X, Y$ this is an integral weighted by the joint PDF:
$$ R(\bbP_{X,Y}, \bfW) = \int_{x \in \Omega_X} \int_{y \in \Omega_Y} loss(\hat{y}, y)\, f_{X,Y}(x, y)\, dx\, dy, \qquad \hat{y} = f(x; \bfW).$$

We can compute the empirical risk, but the expected risk is what we actually want to minimize.

### Q18. 

Why do we split a dataset into training, validation and test data? What is each part used for? What is *data contamination*?

**Answer**

We can never compute the expected risk $R(\bbP_{X,Y}, \bfW) = \bbE_{X,Y} [ loss(\hat{Y}, Y) ]$, so we settle for the next best thing: held-out data.

1. **Training data** $\calD_t$: to train the model by gradient descent or other algorithms.
2. **Validation data** $\calD_v$: to choose hyper-parameters, detect **overfitting**, and decide when to stop training.
3. **Test/evaluation data** $\calD_e$: only for evaluating and comparing algorithms. The test data is sacred.

If training/validation data contains any information from the test data, the training data is **contaminated**. This is a big problem for Large Language Models: companies crawl the internet for training data and often end up including the benchmark answers.

If the three splits are sampled uniformly from the collected dataset $\calD$, they are likely to be identically distributed. Then the test performance is a good estimate of real-world performance.

### Q19. 

Write the gradient descent algorithm for minimizing the average training loss $\calL(\calD_t; \bfW)$.

**Answer**

**given** a starting point $\bfW \in \text{dom}{f}$.

**repeat**
1. Choose step size $\alpha_t$.
2. Update:
   $$\bfW \leftarrow \bfW  - \alpha_t \nabla_\bfW \underbrace{ \left( \frac{1}{|\calD_t|} \sum_{(\bfx_i, \bfy_i) \in \calD_t} l(\bfy_i, f(\bfx_i; \bfW)) \right) }_{\calL(\calD_t; \bfW)}
   = \bfW  -  \frac{\alpha_t }{|\calD_t|} \sum_{(\bfx_i, \bfy_i) \in \calD_t} \nabla_\bfW  l(\bfy_i, f(\bfx_i; \bfW)). $$

**until** the stopping criterion is satisfied.

### Q20. 

Write the Batch Stochastic Gradient Descent (Batch-SGD) algorithm. What is the algorithm called when the batch size $|\calB_j| = 1$?

**Answer**

**given** a starting point $\bfW \in \text{dom}{f}$.

**for** e **in** range(n_epochs):

1. Choose step size $\alpha_t$.
2. Shuffle the training dataset $\calD_t$.
3. For each small batch $\calB_j \subset \calD_t$:
   $$\bfW \leftarrow \bfW  -  \frac{\alpha_t }{|\calB_j|} \sum_{(\bfx_i, \bfy_i) \in \calB_j} \nabla_\bfW l(\bfy_i, f(\bfx_i; \bfW)). $$
4. Evaluate the validation loss $\frac{1}{|\calD_v|}\sum_{(\bfx_i, \bfy_i) \in \calD_v} l(\bfy_i, f(\bfx_i; \bfW))$. Stop when it stops decreasing.

With batch size $|\calB_j|=1$, the algorithm is called Stochastic Gradient Descent (SGD).

### Q21. 

What are the advantages of Batch-SGD over gradient descent (GD)? When should you use plain GD?

**Answer**

**Advantages of Batch-SGD:**

1. **Faster on big datasets.** GD can make one update only after looping over the entire dataset. A uniformly sampled small batch gives roughly the same information as the full dataset, so SGD makes progress much sooner.
2. **Better aligned with the ML objective.** In ML the true objective is the *expected* loss, not the *empirical* training loss. The random choice of small batches keeps Batch-SGD from overfitting to the training set.

**Use GD:**

1. For optimization problems that are not machine learning.
2. With small datasets and simple models, when you know the model you want to fit (e.g. polynomial regression).

### Q22. 

You are doing 0-1 binary classification on MNIST, where detecting the digit 1 is the positive class. Using this example, define False negative, False positive, Accuracy, Precision, Recall and F1-score.

**Answer**

- **False negative:** the classifier predicts negative (0) but the digit is actually 1.
- **False positive:** the classifier predicts positive (1) but the digit is actually 0.
- **Accuracy** = (number of correct classifications) / (total number of samples).
- **Precision** = true positives / predicted positives = $\frac{TP}{TP + FP}$.
- **Recall** = true positives / actual positives = $\frac{TP}{TP + FN}$.
- **F1-score** = harmonic mean of precision and recall = $\frac{2 \cdot \text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}$.

### Q23. 

What is a ROC (Receiver Operating Characteristic) curve for a binary classifier? Argue why it lies above the diagonal for any reasonable classifier. Also argue why it must be concave.

**Answer**

The ROC curve plots the true positive rate $TPR = P(\hat{Y}(X) = 1| Y = 1)$ against the false positive rate $FPR = P(\hat{Y}(X) = 1| Y=0)$ as the decision threshold varies.

**Above the diagonal.** A trivial classifier can predict the positive class with probability $\alpha \in (0, 1)$ without looking at $X$. It has $FPR = TPR = \alpha$, which traces the diagonal. Any classifier better than this trivial one lies above the diagonal.

**Concave.** Take two operating points on the ROC curve: classifier $\hat{Y}_1$ with $FPR = f_1$, $TPR = TPR(f_1)$ and classifier $\hat{Y}_2$ with $FPR = f_2$, $TPR = TPR(f_2)$. Build a randomized classifier $\hat{Y}$ that uses $\hat{Y}_1$ with probability $\alpha$ and $\hat{Y}_2$ with probability $1-\alpha$. Then

\begin{align}
P(\hat{Y}(X) = 1 | Y = 1) &= P(\hat{Y}_1(X) = 1 | Y = 1)\alpha + P(\hat{Y}_2(X) = 1 |  Y = 1)(1-\alpha) \\
&= \alpha\, TPR(f_1) + (1-\alpha)\, TPR(f_2),
\end{align}

and similarly its $FPR = \alpha f_1 + (1-\alpha) f_2$. So any point on the chord between two ROC points is achievable. The ROC curve is the best achievable TPR at each FPR, so it lies on or above every chord:
$$TPR(\alpha  f_1 + (1-\alpha) f_2) \ge \alpha\, TPR(f_1) + (1-\alpha)\, TPR(f_2).$$

That is the definition of concavity: $g$ is concave if $g(\alpha y_1 + (1-\alpha) y_2) \ge \alpha g(y_1) + (1-\alpha) g(y_2)$ for all $y_1, y_2$ and $\alpha \in (0, 1)$.

### Q24. 

What is the likelihood ratio test for a binary classifier? Is it better or worse than the posterior ratio test?

**Answer**

The likelihood is the probability of the evidence $\bfx$ given a class $y \in \{0, 1\}$: $\bbP(\bfx | y)$. The likelihood ratio test predicts the positive class when the likelihood ratio exceeds a threshold $\eta$:
$$
\hat{y}(\bfx) = \begin{cases}
1 & \text{ if } \frac{\bbP(\bfx | y = 1)}{\bbP(\bfx | y = 0)} \ge \eta \\
0 & \text{ otherwise }
\end{cases}
$$

The posterior is $\bbP(y | \bfx) = \frac{\bbP(\bfx | y) \bbP(y)}{\bbP(\bfx)}$, and the posterior ratio test is
$$
\hat{y}(\bfx) = \begin{cases}
1 & \text{ if } \frac{\bbP(y = 1 | \bfx)}{\bbP(y = 0| \bfx)} \ge \eta_p \\
0 & \text{ otherwise }
\end{cases}
$$

**Neither is better: they are equivalent.** The posterior ratio condition can be rewritten as a likelihood ratio condition:
\begin{align}
\frac{\bbP(y = 1 | \bfx)}{\bbP(y = 0| \bfx)} \ge \eta_p
\iff
\frac{\bbP(\bfx | y = 1)\bbP(y=1)}{\bbP( \bfx|y=0)\bbP(y=0)} \ge \eta_p
\iff
\frac{\bbP(\bfx | y = 1)}{\bbP( \bfx|y=0)} \ge \frac{\bbP(y=0)}{\bbP(y=1)}\eta_p.
\end{align}

This is the likelihood ratio test with $\eta = \frac{\bbP(y=0)}{\bbP(y=1)}\eta_p$.

## Part B: Automatic differentiation with scalars

### Three ways to compute derivatives on a computer. Briefly compare them.

**Answer**

1. **Numerical differentiation:** finite differences, e.g. $f'(x) \approx \frac{f(x+h) - f(x)}{h}$. Easy, but approximate (truncation and round-off error) and needs one extra function evaluation per input dimension.
2. **Symbolic differentiation:** manipulate the expression to get an exact formula for the derivative (as in Mathematica/SymPy). Exact, but the expressions can blow up in size and it does not work directly on arbitrary programs (loops, branches).
3. **Automatic differentiation:** apply the chain rule to the elementary operations that the program actually executes. Exact up to floating point and efficient. It has two modes:
    1. Forward mode: propagates derivatives from the input side to the output side, alongside the values.
    2. Reverse mode (backpropagation): first computes the values, then propagates derivatives from the output side back to the inputs.

### Q32. (Will not be asked in exam)

Derive the scalar single-variable chain rule $\frac{\p f(g(x))}{\p x} = f'(g(x))g'(x)$ from the limit definition of the derivative.

**Answer**

From $g'(x) = \lim_{h \to 0} \frac{g(x+h) - g(x)}{h}$ we get $\lim_{h \to 0} g(x+h) = \lim_{h \to 0} g(x) + g'(x)h$. Then
\begin{align}
\frac{\p f(g(x))}{\p x} &= \lim_{h \to 0} \frac{f(g(x + h)) - f(g(x))}{h}\\
&=\lim_{h \to 0} \frac{f(g(x) + g'(x) h) - f(g(x))}{h}\\
&=\lim_{h \to 0} \frac{f(g(x)) + f'(g(x))g'(x)h - f(g(x))}{h}\\
&=f'(g(x))g'(x).
\end{align}

### Q33. (Will not be asked in exam)

Derive the chain rule for a function of two variables, 

$\frac{\p}{\p x} f(u(x), v(x)) = ?$.

**Answer**

\begin{align}
\frac{\p f(u(x), v(x))}{\p x} &= \lim_{h \to 0} \frac{f(u(x + h), v(x+h)) - f(u(x), v(x))}{h}
=\lim_{h \to 0} \frac{f(u + u'(x) h, v + v'(x)h) - f(u, v)}{h}.
\end{align}

Expand $f(u + \delta u, v + \delta v)$ in two steps, first in $u$ and then in $v$:
$$f(u+\delta u, v+\delta v) \approx f(u, v+\delta v) + f'_u(u, v+\delta v) \delta u \approx f(u, v) + f'_v(u, v) \delta v + f'_u(u, v+\delta v) \delta u.$$

Since $f'_u(u, v+\delta v)\delta u = f'_u(u, v)\delta u + f''_{uv}(u, v)\,\delta v\, \delta u$ and the second-order term vanishes in the limit,
$$f(u+\delta u, v+\delta v) \approx f(u, v) + f'_v(u, v) \delta v + f'_u(u, v) \delta u.$$

With $\delta u = u'(x)h$ and $\delta v = v'(x)h$:
$$\frac{\p f(u(x), v(x))}{\p x} = f'_u(u(x), v(x))\, u'(x) + f'_v(u(x), v(x))\, v'(x).$$

### Q34. (Will not be asked in exam)

Let $f(\bfg): \bbR^m \to \bbR$ and $\bfg(x): \bbR \to \bbR^m$. Using the row-vector (numerator-layout) convention, what are the shapes of $\frac{\p f}{\p \bfg}$ and $\frac{\p \bfg}{\p x}$? Write the chain rule for $\frac{\p}{\p x} f(\bfg(x))$. Does the order of multiplication matter?

**Answer**

$\frac{\p f}{\p \bfg} = \begin{bmatrix} \frac{\p f}{\p g_1} & \dots & \frac{\p f}{\p g_m} \end{bmatrix}$ is a $1 \times m$ row vector. $\frac{\p \bfg}{\p x} = \begin{bmatrix} \frac{\p g_1}{\p x} & \dots & \frac{\p g_m}{\p x} \end{bmatrix}^\top$ is an $m \times 1$ column vector.

$f(\bfg(x)) = f(g_1(x), \dots, g_m(x))$ is a multivariate scalar function, so by the multivariable chain rule
$$\frac{\p}{\p x} f(\bfg(x)) = f'_{g_1}(\bfg(x)) g'_1(x) + \dots + f'_{g_m}(\bfg(x)) g'_m(x)
= \begin{bmatrix} f'_{g_1} & \dots & f'_{g_m} \end{bmatrix}
\begin{bmatrix} g'_1(x) \\ \vdots \\  g'_m(x) \end{bmatrix}
= \frac{\p f}{\p \bfg}\frac{\p \bfg}{\p x}.$$

The order matters: $\frac{\p \bfg}{\p x} \frac{\p f}{\p \bfg}$ would be an $m \times m$ outer product, not the derivative. With a column-vector convention the order would be reversed.

### Q36. 

How many floating point operations does it take to multiply $A \in \bbR^{m \times n}$ by $B \in \bbR^{n \times p}$?

$C = AB$ has $mp$ entries. Each entry is a dot product of a row of $A$ and a column of $B$, $\bfa_i^\top \bfb_j = a_{i1} b_{1j} + \dots + a_{in} b_{nj}$, which takes $n$ multiplications and $n-1$ additions.

### Q37. 

Consider $\bfh(\bfx): \bbR^m \to \bbR^n$, $\bfg(\bfh): \bbR^n \to \bbR^p$ and $\bff(\bfg): \bbR^p \to \bbR^q$, so that
$$ \frac{\p }{\p \bfx} \bff(\bfg(\bfh(\bfx))) = \frac{\p \bff}{\p \bfg} \frac{\p \bfg}{\p \bfh} \frac{\p \bfh}{\p \bfx}. $$

(a) What is the computational complexity of forward-mode and of reverse-mode differentiation?
(b) Which is cheaper for $m = 1000$, $n = 100$, $p = 10$, $q = 1$?
(c) What happens for a long chain of $k$ functions?

**Answer**

The Jacobians have shapes $\frac{\p \bff}{\p \bfg}: q \times p$, $\frac{\p \bfg}{\p \bfh}: p \times n$ and $\frac{\p \bfh}{\p \bfx}: n \times m$.

(a) **Forward mode** multiplies from the input side (right to left):
$$\frac{\p \bff}{\p \bfg} \left( \frac{\p \bfg}{\p \bfh} \frac{\p \bfh}{\p \bfx} \right): \quad O(pnm) + O(qpm) = O\big((pn + qp)\,m\big).$$

**Reverse mode** multiplies from the output side (left to right):
$$\left(\frac{\p \bff}{\p \bfg}  \frac{\p \bfg}{\p \bfh}\right) \frac{\p \bfh}{\p \bfx}: \quad O(qpn) + O(qnm) = O\big(q\,(pn + nm)\big).$$

(b) Forward: $10\cdot100\cdot1000 + 1\cdot10\cdot1000 = 1{,}010{,}000$. Reverse: $1\cdot10\cdot100 + 1\cdot100\cdot1000 = 101{,}000$. Reverse mode is about 10 times cheaper.

(c) For a chain of Jacobians $J_k \cdots J_2 J_1$ where $J_i$ has shape $q_i \times p_i$ ($p_i = q_{i-1}$), forward mode costs $O\big((\sum_{i\ge2} q_i p_i)\, p_1\big)$: the input size $p_1$ multiplies every term. Reverse mode costs $O\big(q_k \sum_{i<k} q_i p_i\big)$: the output size $q_k$ multiplies every term. So forward mode is cheaper when there are few inputs, and reverse mode is cheaper when there are few outputs.

### Q38. 

Why is reverse-mode differentiation (backpropagation) the standard method for training neural networks?

**Answer**

We differentiate the loss function, which is a scalar: its output dimension is $q = 1$. Reverse-mode cost is proportional to the output dimension, and forward-mode cost is proportional to the input dimension (the number of weights, often millions). One reverse pass gives the gradient with respect to all weights, so reverse mode is clearly superior for loss-function gradients. In neural networks, reverse-mode differentiation is called backpropagation.

### Q39. 

What is a vector-Jacobian product (VJP)? Write the template of a `vjp` function for a function `g(arg1, arg2)` used in reverse mode. Why is it called a VJP, and why don't we always compute the Jacobian explicitly?

**Answer**

The output of the network is a scalar loss $l$, so the left-most factor in the reverse-mode chain rule is a row vector:
$$ \frac{\p }{\p \bfx} l(\bff(\bfg(\bfx))) = \frac{\p l}{\p \bff}\frac{\p \bff}{\p \bfg}\frac{\p \bfg}{\p \bfx}.$$

A `vjp` function for $\bfg$ receives the accumulated row vector $\frac{\p l}{\p \bfg}$ and returns $\frac{\p l}{\p \bfg}\frac{\p \bfg}{\p \text{arg}}$ for each argument: a row **vector** times the local **Jacobian**.

It helps to imagine a scalar loss $l(\bfg(\bfx))$ whose gradient $\frac{\p l}{\p \bfg}$ is given. You don't have to build the Jacobian separately: it is often large and sparse (e.g. diagonal for element-wise operations), and the product can be computed much more cheaply directly.

### Q40. 

What is a Jacobian-vector product (JVP) used in forward mode? Write the template of a `jvp` function. How do you compute $\frac{\p \bff}{\p x_2}$ in forward mode, and how many passes does each mode need to compute the full Jacobian of $\bff: \bbR^m \to \bbR^q$?

**Answer**

Assume the input is a function of a scalar $t$, $\bfx(t)$. Then
$$ \frac{\p }{\p t} \bff(\bfg(\bfh(\bfx(t)))) = \frac{\p \bff}{\p \bfg} \frac{\p \bfg}{\p \bfh} \frac{\p \bfh}{\p \bfx} \frac{\p \bfx}{\p t}. $$

A `jvp` function multiplies the local Jacobian by the incoming column vector of derivatives with respect to $t$:

To get $\frac{\p \bff}{\p x_2}$, seed $\frac{\p \bfx}{\p t} = \bfe_2 = [0, 1, 0, \dots, 0]^\top$.

Forward mode gives one **column** of the Jacobian per pass, so it needs $m$ passes (one per input). Reverse mode gives one **row** per pass, so it needs $q$ passes (one per output).

Find the VJP of element-wise addition $\bff(\bfa, \bfb) = \bfa + \bfb$, where $\bfa, \bfb, \bff \in \bbR^n$. That is, given $\frac{\p l}{\p \bff}$, find $\frac{\p l}{\p \bfa}$ and $\frac{\p l}{\p \bfb}$.

### Q46. 

Find the VJP of matrix-vector multiplication $\bff(\bfA, \bfb) = \bfA \bfb$, where $\bff \in \bbR^m$, $\bfb \in \bbR^n$ and $\bfA \in \bbR^{m \times n}$. Define $\frac{\p l}{\p \bfA}$ to have the same shape as $\bfA$, with entries $\frac{\p l}{\p a_{ij}}$.

**Answer**

Write $\bfA$ in terms of its rows $\bfa_i^\top$, so that $\bfA\bfb = \begin{bmatrix} \bfa_1^\top\bfb & \dots & \bfa_m^\top\bfb \end{bmatrix}^\top$.

Each $f_i = \bfa_i^\top \bfb$ is a scalar that depends only on row $i$ of $\bfA$. So $\frac{\p f_i}{\p \bfA}$ is an $m \times n$ matrix whose $i$-th row is $\bfb^\top$ and whose other rows are zero. With $\frac{\p l}{\p \bff} = \begin{bmatrix} \frac{\p l}{\p f_1} & \dots & \frac{\p l}{\p f_m} \end{bmatrix}$, row $i$ of $\frac{\p l}{\p \bfA}$ is $\frac{\p l}{\p f_i}\bfb^\top$:

$$\frac{\p l}{\p \bfA} =
\begin{bmatrix}
\frac{\p l}{\p f_1}\bfb^\top\\
\vdots\\
\frac{\p l}{\p f_m}\bfb^\top
\end{bmatrix}
= \left(\frac{\p l}{\p \bff}\right)^\top \bfb^\top = \left(\bfb\frac{\p l}{\p \bff}\right)^\top \in \bbR^{m \times n}.$$

The derivative with respect to $\bfb$ is simpler:
$$\frac{\p l}{\p \bfb} = \frac{\p l}{\p \bff}\frac{\p }{\p \bfb}(\bfA\bfb) = \frac{\p l}{\p \bff} \bfA \in \bbR^{1 \times n}.$$

### Q47. 

Find the VJP of matrix-matrix multiplication $\bfF(\bfA, \bfB) = \bfA \bfB$, where $\bfF \in \bbR^{m \times p}$, $\bfA \in \bbR^{m \times n}$ and $\bfB \in \bbR^{n \times p}$. Check your answer by matching shapes.

**Answer**

Write $\bfB$ and $\bfF$ in terms of their columns: $\bff_i = \bfA\bfb_i$ for $i = 1, \dots, p$. From the matrix-vector VJP, $\frac{\p l}{\p \bff_i}\frac{\p \bff_i}{\p \bfA} = \left(\bfb_i\frac{\p l}{\p \bff_i}\right)^\top$. Since $l$ depends on $\bfA$ through all the columns $\bff_1, \dots, \bff_p$, the multivariable chain rule sums over them:

$$\frac{\p l}{\p \bfA} = \sum_{i=1}^p \left(\bfb_i\frac{\p l}{\p \bff_i}\right)^\top
= \left(\begin{bmatrix} \bfb_1 & \dots & \bfb_p \end{bmatrix}
\begin{bmatrix} \frac{\p l}{\p \bff_1}\\ \vdots\\ \frac{\p l}{\p \bff_p}\end{bmatrix}\right)^\top
= \left(\bfB \left(\frac{\p l}{\p \bfF}\right)^\top\right)^\top
= \frac{\p l}{\p \bfF}\bfB^\top.$$

For $\bfB$, apply the same rule to $\bfF^\top = \bfB^\top \bfA^\top$ and transpose:
$$ \frac{\p l}{\p \bfB} = \bfA^\top\frac{\p l}{\p \bfF}. $$

**Shape check:** $\frac{\p l}{\p \bfF}$ is $m \times p$, so $\frac{\p l}{\p \bfF}\bfB^\top$ is $(m \times p)(p \times n) = m \times n$, the shape of $\bfA$. $\bfA^\top\frac{\p l}{\p \bfF}$ is $(n \times m)(m \times p) = n \times p$, the shape of $\bfB$. Matching shapes is a quick way to remember these formulas.

### Q48. 

Write the `vjp` functions for the element-wise operations `exp`, `log`, `maximum(a, b)` (ReLU is `maximum(x, 0)`), and for `sum(x, axis)`. All functions take the incoming gradient `dldf` as the first argument.

**Answer**

- $\frac{d}{dx} e^x = e^x$, $\frac{d}{dx} \log x = 1/x$.
- `maximum` passes the gradient to whichever argument is larger.
- `sum` copies the incoming gradient to every summed element, so it is broadcast back to the shape of `x`.

````python
# Code not to be written in exam
import numpy as np

def exp_vjp(dldf, x):
    return (unbroadcast(x, dldf * np.exp(x)),)

def log_vjp(dldf, x):
    return (unbroadcast(x, dldf / x),)

def maximum_vjp(dldf, a, b):
    dlda = dldf * np.where(a > b, 1, 0)
    dldb = dldf * np.where(a > b, 0, 1)
    return unbroadcast(a, dlda), unbroadcast(b, dldb)

def sum_vjp(dldf, x, axis=None, **kwargs):
    if axis is not None:
        dldx = np.expand_dims(dldf, axis=axis) * np.ones_like(x)
    else:
        dldx = dldf * np.ones_like(x)
    return (unbroadcast(x, dldx),)
````

### Q49. (Not for exam)

What is the `unbroadcast` function for? Suppose `A.shape = (100, 1)`, `B.shape = (1, 100)` and `F = A + B`. What are the shapes of `dl__dF`, `dl__dA` and `dl__dB`, and how do you compute `dl__dA` and `dl__dB` from `dl__dF`? Also, what will the code below print?

````python
import numpy as np

def unbroadcast(target, g, axis=0):
    """Remove broadcasted dimensions by summing along them."""
    while np.ndim(g) > np.ndim(target):
        g = g.sum(axis=axis)
    for axis, size in enumerate(target.shape):
        if size == 1:
            g = g.sum(axis=axis, keepdims=True)
    return g

g = unbroadcast(np.zeros((3, 1)), np.ones((2, 3, 4)))
print(g.shape)
print(g.ravel())
````

**Answer**

`F.shape = (100, 100)` because [NumPy broadcasting](https://numpy.org/doc/stable/user/basics.broadcasting.html) copies `A` 100 times along axis 1 and `B` 100 times along axis 0. So `dl__dF.shape = (100, 100)`, but the gradients must match their inputs: `dl__dA.shape = (100, 1)` and `dl__dB.shape = (1, 100)`.

**Why sum?** Consider $f(y_1, y_2) = y_1 y_2$ where $y_1(x) = x$ and $y_2(x) = x$ are copies of the same variable. By the chain rule,
$$\p_x f = \frac{\p f}{\p y_1}\p_x y_1 + \frac{\p f}{\p y_2}\p_x y_2 = \frac{\p f}{\p y_1} + \frac{\p f}{\p y_2}.$$
When a variable is copied and the copies are used in a function, the derivative is the **sum** of the derivatives with respect to the copies. Broadcasting creates copies along the mismatched axes, so `unbroadcast` sums the gradient along those axes:

```python
dl__dA = dl__dF.sum(axis=1, keepdims=True)  # (100, 1)
dl__dB = dl__dF.sum(axis=0, keepdims=True)  # (1, 100)
```

The code prints
```
(3, 1)
[8. 8. 8.]
```
The extra leading axis is summed first, giving shape `(3, 4)` with value 2. Then axis 1 (size 1 in the target) is summed with `keepdims`, giving shape `(3, 1)` with value $2 \times 4 = 8$.
