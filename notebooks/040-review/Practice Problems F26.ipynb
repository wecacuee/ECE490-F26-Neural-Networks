A10. The mangitude of n-D vector is given by $\|\bfa\| = \sqrt{a_1^2 + a_2^2 + \dots + a_n^2}$ and dot product the vector with itself is given by

\end{align}
If $\bfy$ is a function of $\bfx$, then
\begin{align}
\frac{\p }{\p \bfx} \bfx^\top \bfy &= \bfy^\top + 
\left(\frac{\p }{\p \bfy} \bfx^\top \bfy \right) \left(\frac{\p \bfy }{\p \bfx}  \right)\\
&= \bfy^\top + \bfx^\top \left(\frac{\p \bfy }{\p \bfx}\right)
\end{align}
If $\bfy = \bfA \bfx$, then 
$$\frac{\p \bfy}{\p \bfx} = \frac{\p }{\p \bfx} \bfA \bfx = \bfA$$
and 

$$\frac{\p }{\p \bfx} \bfx^\top \bfA \bfx = 
\bfy^\top + \bfx^\top \left(\frac{\p \bfy }{\p \bfx}\right)
= \bfx^\top \bfA^\top + \bfx^\top \bfA 
= \bfx^\top (\bfA^\top + \bfA)
$$

### Q26 Product of two Gaussian distributions (20 marks)

$$ f(\bfx; \mu, \Sigma) = \frac{1}{\sqrt{(2\pi)^n  \det(\Sigma)})} \exp\left(
 -\frac{1}{2}(\bfx - \mu)^\top \Sigma^{-1} (\bfx - \mu) \right),$$

 where the mean $\mu \in \bbR^n$ is an n-D vector and $\Sigma \in \bbR^{n\times n}$ is an $n\times n$ square invertible and positive definite covariance matrix.

 In this problem, you have to find the mode (equivalently mean) of the Gaussian distribution when two distributions are multiplied together:
 
 $$\mu = \arg~\max_{\bfx} f(\bfx; \mu_1, \Sigma_1)f(\bfx; \mu_2, \Sigma_2)$$

HINT 1: Maximizing the negative exponent $\exp(-l(\bfx))$ of a function $l(\bfx)$  is same as minmimizing the function $l(\bfx)$.

This is same as minimizing $$ \arg~\min_{\bfx} \; (\bfx - \bfmu_1)^\top \Sigma_1^{-1} (\bfx - \bfmu_1) + (\bfx - \bfmu_2)^\top \Sigma_2^{-1} (\bfx - \bfmu_2)  $$
