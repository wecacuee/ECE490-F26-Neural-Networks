## What is a Neural Network?


<img src="imgs/parts-of-a-neuron.webp" width="500px" alt="a neuron with dendrite, soma, axon, myelin, synapse, axon terminal"/>

<cite data-cite="kirby2024openstax">(Kirby "Introduction to Behavioral Neuroscience", 2024; OpenStax)</cite>

"The human brain is made up of ~86 billion neurons. Neurons are highly specialized for communication that occurs via electrical and chemical signals. With some exceptions, neurons have special structural features that branch from the cell body (soma), called axons and dendrites. Functionally, neurons receive and integrate signals, and send information in both electrical and chemical forms. Neurons communicate with their partners at regions called synapses."

"A neuron can have thousands of dendrites, but only one axon. An axon is a branch that like dendrites, emerges from the soma or cell body of the neuron."
