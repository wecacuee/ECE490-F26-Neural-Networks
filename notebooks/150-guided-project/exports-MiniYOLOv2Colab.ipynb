```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import numpy as np
import os
from PIL import Image

# ── Architecture constants ────────────────────────────────────────────────────
REG_MAX = 8   # DFL bins: each box param is distributed over REG_MAX discrete values

# ── Building blocks ───────────────────────────────────────────────────────────

class ConvBNAct(nn.Module):
    """Conv → BN → SiLU (modern activation used in YOLOv8+)."""
    def __init__(self, in_c, out_c, k=3, s=1, p=1):
        super().__init__()
        self.block = nn.Sequential(
            nn.Conv2d(in_c, out_c, k, s, p, bias=False),
            nn.BatchNorm2d(out_c),
            nn.SiLU(inplace=True),
        )
    def forward(self, x):
        return self.block(x)


class ResBlock(nn.Module):
    """Two 3×3 convs with an identity skip connection."""
    def __init__(self, c):
        super().__init__()
        self.cv1 = ConvBNAct(c, c, 3)
        self.cv2 = ConvBNAct(c, c, 3)
    def forward(self, x):
        return x + self.cv2(self.cv1(x))


class CSPBlock(nn.Module):
    """Cross Stage Partial: split channels → residual branch + bypass → fuse.
    Improves gradient flow while keeping parameter count low."""
    def __init__(self, in_c, out_c, n=1):
        super().__init__()
        mid = out_c // 2
        self.cv1 = ConvBNAct(in_c, mid, 1, p=0)
        self.cv2 = ConvBNAct(in_c, mid, 1, p=0)
        self.blocks = nn.Sequential(*[ResBlock(mid) for _ in range(n)])
        self.cv3 = ConvBNAct(mid * 2, out_c, 1, p=0)

    def forward(self, x):
        return self.cv3(torch.cat([self.blocks(self.cv1(x)), self.cv2(x)], dim=1))

# ── Model ─────────────────────────────────────────────────────────────────────

class MiniYOLOv2(nn.Module):
    """
    Backbone : stem + 4 CSP stages with strided downsamples.
    Neck     : FPN top-down skip from P4 fused into P5, then bottom-up merge.
    Head     : 1×1 conv outputting per-cell:
                 [obj | l_bins(R) | t_bins(R) | r_bins(R) | b_bins(R) | cls(C)]

    DFL encoding (all scaled to [0, R-1]):
        d0 = x_rel * (R-1)   d1 = y_rel * (R-1)
        d2 = w_norm * G       d3 = h_norm * G
    """
    def __init__(self, num_classes=80, img_size=256, reg_max=REG_MAX):
        super().__init__()
        self.img_size    = img_size
        self.grid_size   = img_size // 32
        self.num_classes = num_classes
        self.reg_max     = reg_max

        self.stem   = ConvBNAct(3,  16, 3, s=2)
        self.stage1 = nn.Sequential(CSPBlock(16, 32, n=1), ConvBNAct(32, 32, 3, s=2))
        self.stage2 = nn.Sequential(CSPBlock(32, 64, n=1), ConvBNAct(64, 64, 3, s=2))  # P3
        self.stage3 = nn.Sequential(CSPBlock(64, 64, n=2), ConvBNAct(64, 64, 3, s=2))  # P4
        self.stage4 = nn.Sequential(CSPBlock(64, 64, n=1), ConvBNAct(64, 64, 3, s=2))  # P5

        self.up         = nn.Upsample(scale_factor=2, mode='nearest')
        self.neck_merge = CSPBlock(128, 64, n=1)   # P5↑ cat P4 → /16
        self.neck_down  = ConvBNAct(64, 64, 3, s=2)
        self.neck_fuse  = CSPBlock(128, 64, n=1)   # neck_down cat P5 → /32

        self.head = nn.Conv2d(64, 1 + 4 * reg_max + num_classes, 1)

    def forward(self, x):
        x  = self.stem(x)
        x  = self.stage1(x)
        x  = self.stage2(x)
        p4 = self.stage3(x)
        p5 = self.stage4(p4)
        m   = self.neck_merge(torch.cat([self.up(p5), p4], dim=1))
        d   = self.neck_down(m)
        out = self.neck_fuse(torch.cat([d, p5], dim=1))
        out = self.head(out)
        return out.permute(0, 2, 3, 1).contiguous()   # [B, G, G, 1+4R+C]


def count_parameters(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

model = MiniYOLOv2(num_classes=2)
print(f"Total trainable parameters: {count_parameters(model)}")


# ── Loss: BCE (obj/cls) + Distribution Focal Loss (box) ──────────────────────

class YOLOLossV2(nn.Module):
    """
    YOLO loss with DFL for box regression.
      Objectness : BCEWithLogitsLoss (all cells, no-obj weighted by lambda_noobj)
      Box        : DFL soft cross-entropy over REG_MAX bins (obj cells only)
      Class      : BCEWithLogitsLoss (obj cells only)

    All terms are normalised by the number of positive (object) cells so the
    loss scale stays consistent regardless of batch size or image density.
    """
    def __init__(self, reg_max=REG_MAX, lambda_coord=5.0, lambda_noobj=0.5):
        super().__init__()
        self.reg_max      = reg_max
        self.lambda_coord = lambda_coord
        self.lambda_noobj = lambda_noobj
        self.bce = nn.BCEWithLogitsLoss(reduction='none')

    def _encode_dfl_targets(self, targets):
        R, G = self.reg_max, targets.shape[1]
        x = targets[..., 1] * (R - 1)
        y = targets[..., 2] * (R - 1)
        w = targets[..., 3] * G
        h = targets[..., 4] * G
        return torch.stack([x, y, w, h], dim=-1).clamp(0, R - 1 - 1e-4)

    def _dfl_loss(self, pred_dist, target_d):
        """Soft cross-entropy: split target between neighbouring bins."""
        R = self.reg_max
        pred_flat = pred_dist.reshape(-1, R)
        t  = target_d.reshape(-1)
        tl = t.long().clamp(0, R - 2)
        tr = tl + 1
        wl = (tr.float() - t).clamp(0.0, 1.0)
        wr = 1.0 - wl
        loss = (F.cross_entropy(pred_flat, tl, reduction='none') * wl
              + F.cross_entropy(pred_flat, tr, reduction='none') * wr)
        return loss.reshape(-1, 4).sum(-1)   # [N]

    def forward(self, preds, targets):
        R          = self.reg_max
        obj_mask   = targets[..., 0] == 1
        noobj_mask = ~obj_mask
        # Normalise all terms by the number of objects in the batch
        n_obj = obj_mask.sum().clamp(min=1).float()

        # Objectness BCE (all cells)
        conf_loss       = self.bce(preds[..., 0], targets[..., 0])
        obj_conf_loss   = (obj_mask.float()   * conf_loss).sum() / n_obj
        noobj_conf_loss = (noobj_mask.float() * conf_loss).sum() / n_obj

        # DFL box loss (obj cells only)
        if obj_mask.any():
            pred_dist  = preds[..., 1:1 + 4 * R][obj_mask].reshape(-1, 4, R)
            target_dfl = self._encode_dfl_targets(targets)[obj_mask]
            box_loss   = self._dfl_loss(pred_dist, target_dfl).sum() / n_obj
        else:
            box_loss = preds.new_tensor(0.0)

        # Class BCE (obj cells only)
        cls_loss = self.bce(preds[..., 1 + 4 * R:], targets[..., 5:])
        cls_loss = (obj_mask.unsqueeze(-1).float() * cls_loss).sum() / n_obj

        return (self.lambda_coord * box_loss
                + obj_conf_loss
                + self.lambda_noobj * noobj_conf_loss
                + cls_loss)
```

```python
import torch
import numpy as np
from PIL import Image
import os

class YOLODataset(torch.utils.data.Dataset):
    def __init__(self, root_dir, set_name, img_size=256, grid_size=8, num_classes=2, transform=None, max_images=None):
        self.root_dir = root_dir
        self.set_name = set_name
        self.img_dir = os.path.join(root_dir, 'images', set_name)
        self.label_dir = os.path.join(root_dir, 'labels', set_name)
        self.img_size = img_size
        self.grid_size = grid_size
        self.num_classes = num_classes
        self.transform = transform

        all_image_files = [f for f in os.listdir(self.img_dir) if f.endswith(('.jpg', '.jpeg', '.png'))]
        all_image_files.sort() # Ensure consistent order

        if max_images is not None:
            self.image_files = all_image_files[:max_images]
        else:
            self.image_files = all_image_files

    def __len__(self):
        return len(self.image_files)

    def __getitem__(self, idx):
        img_name = self.image_files[idx]
        img_path = os.path.join(self.img_dir, img_name)
        label_path = os.path.join(self.label_dir, img_name.replace('.jpg', '.txt').replace('.jpeg', '.txt').replace('.png', '.txt'))

        # Load image
        img = Image.open(img_path).convert('RGB')
        img = img.resize((self.img_size, self.img_size))
        img = np.array(img).astype(np.float32) / 255.0 # Normalize to [0, 1]

        # Initialize target tensor for YOLO format (grid_size, grid_size, 5 + num_classes)
        # Each grid cell will store: [objectness, x_center_rel, y_center_rel, width_rel, height_rel, class_probs...]
        # x_center_rel, y_center_rel, width_rel, height_rel are relative to the grid cell.
        target = torch.zeros((self.grid_size, self.grid_size, 5 + self.num_classes))

        if os.path.exists(label_path):
            with open(label_path, 'r') as f:
                for line in f.readlines():
                    parts = line.strip().split()
                    class_id = int(parts[0])
                    x_c_norm, y_c_norm, w_norm, h_norm = map(float, parts[1:])

                    # Denormalize to image coordinates (pixels)
                    x_c_px = x_c_norm * self.img_size
                    y_c_px = y_c_norm * self.img_size

                    # Determine which grid cell the center of the object falls into
                    cell_x = int(self.grid_size * x_c_norm)
                    cell_y = int(self.grid_size * y_c_norm)

                    # Calculate box coordinates relative to the grid cell
                    grid_cell_size_px = self.img_size / self.grid_size
                    x_in_cell_px = x_c_px - (cell_x * grid_cell_size_px)
                    y_in_cell_px = y_c_px - (cell_y * grid_cell_size_px)

                    x_center_rel = x_in_cell_px / grid_cell_size_px # 0 to 1 within the cell
                    y_center_rel = y_in_cell_px / grid_cell_size_px # 0 to 1 within the cell

                    # Set objectness score to 1 and class probabilities
                    if cell_y < self.grid_size and cell_x < self.grid_size: # Ensure cell is within bounds
                        target[cell_y, cell_x, 0] = 1.0  # Objectness score
                        target[cell_y, cell_x, 1:5] = torch.tensor([x_center_rel, y_center_rel, w_norm, h_norm])
                        target[cell_y, cell_x, 5 + class_id] = 1.0 # One-hot encode class

        img = torch.from_numpy(img).permute(2, 0, 1) # Convert to C, H, W format

        if self.transform:
            img = self.transform(img)

        return img, target
```

```python
# --- 3. Functional Training Setup ---

IMG_SIZE    = 256
GRID_SIZE   = IMG_SIZE // 32   # 8×8 grid (stride 32)
NUM_CLASSES = 80

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

dataset = YOLODataset(root_dir=COCO_DATA_DIR, set_name='train2017',
                      img_size=IMG_SIZE, grid_size=GRID_SIZE,
                      num_classes=NUM_CLASSES, max_images=1000)
dataloader = DataLoader(dataset, batch_size=8, shuffle=True, num_workers=0)

model     = MiniYOLOv2(num_classes=NUM_CLASSES, img_size=IMG_SIZE, reg_max=REG_MAX).to(device)
optimizer = optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-4)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=5, eta_min=1e-5)
criterion = YOLOLossV2(reg_max=REG_MAX, lambda_coord=5.0, lambda_noobj=0.5)

print(f"Loaded {len(dataset)} images | Model params: {count_parameters(model):,}")
print(f"Output per cell: 1 (obj) + 4×{REG_MAX} (DFL) + {NUM_CLASSES} (cls) = {1+4*REG_MAX+NUM_CLASSES}")

# Verify one forward + backward pass
images, targets = next(iter(dataloader))
images, targets = images.to(device), targets.to(device)

model.eval()
with torch.no_grad():
    preds     = model(images)
    prev_loss = criterion(preds, targets)

model.train()
optimizer.zero_grad()
preds = model(images)
loss  = criterion(preds, targets)
loss.backward()
optimizer.step()

print(f"Test batch loss: {loss.item():.4f} (before: {prev_loss.item():.4f}) — single iteration works!")
```

```python
# --- 4. Full Training Loop with Validation and LR Scheduling ---
import csv

num_epochs = 5
log_file   = 'training_log_v2.csv'

val_dataset = YOLODataset(root_dir=COCO_DATA_DIR, set_name='val2017',
                          img_size=IMG_SIZE, grid_size=GRID_SIZE,
                          num_classes=NUM_CLASSES, max_images=200)
val_loader = DataLoader(val_dataset, batch_size=8, shuffle=False, num_workers=0)

with open(log_file, mode='w', newline='') as f:
    csv.writer(f).writerow(['epoch', 'train_loss', 'val_loss', 'lr'])

print(f"Starting training for {num_epochs} epochs on {device}...")
train_losses, val_losses = [], []

for epoch in range(num_epochs):
    model.train()
    epoch_train_loss = 0.0
    for batch_idx, (images, targets) in enumerate(dataloader):
        images, targets = images.to(device), targets.to(device)
        optimizer.zero_grad()
        loss = criterion(model(images), targets)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), max_norm=10.0)
        optimizer.step()
        epoch_train_loss += loss.item()
        if (batch_idx + 1) % 50 == 0:
            print(f"  Epoch [{epoch+1}/{num_epochs}] Batch [{batch_idx+1}/{len(dataloader)}]  loss={loss.item():.4f}")

    scheduler.step()
    avg_train = epoch_train_loss / len(dataloader)

    model.eval()
    epoch_val_loss = 0.0
    with torch.no_grad():
        for images, targets in val_loader:
            images, targets = images.to(device), targets.to(device)
            epoch_val_loss += criterion(model(images), targets).item()
    avg_val = epoch_val_loss / len(val_loader)

    current_lr = scheduler.get_last_lr()[0]
    with open(log_file, mode='a', newline='') as f:
        csv.writer(f).writerow([epoch + 1, avg_train, avg_val, current_lr])

    train_losses.append(avg_train)
    val_losses.append(avg_val)
    print(f"==> Epoch {epoch+1}  train={avg_train:.4f}  val={avg_val:.4f}  lr={current_lr:.2e}\n")

print(f"Training complete. Log saved to {log_file}")
```

```python
# --- 5. Visualize Sample Results with Bounding Boxes (DFL decode) ---
import matplotlib.patches as patches

def _cls_label(cls_id):
    """Return class name or fallback string if id is out of range."""
    if 0 <= cls_id < len(COCO_CLASSES):
        return COCO_CLASSES[cls_id]
    return f"cls {cls_id}"

def decode_boxes_v2(pred, img_size, grid_size, reg_max, conf_thresh=0.5):
    """Decode MiniYOLOv2 DFL output [G, G, 1+4R+C] to pixel-space boxes.

    DFL decode per param (weighted sum of bin indices via softmax):
        d0 → x_rel  = weighted_sum / (R-1)
        d1 → y_rel  = weighted_sum / (R-1)
        d2 → w_norm = weighted_sum / G
        d3 → h_norm = weighted_sum / G
    Returns list of (x1, y1, x2, y2, conf, class_id).
    """
    R         = reg_max
    G         = grid_size
    cell_size = img_size / G
    bins      = torch.arange(R, dtype=torch.float32)
    conf      = torch.sigmoid(pred[..., 0])
    boxes     = []
    for gy in range(G):
        for gx in range(G):
            c = conf[gy, gx].item()
            if c < conf_thresh:
                continue
            dist_logits = pred[gy, gx, 1:1 + 4 * R].reshape(4, R)
            d = (torch.softmax(dist_logits, dim=-1) * bins).sum(-1)
            x_rel  = d[0].item() / (R - 1)
            y_rel  = d[1].item() / (R - 1)
            w_norm = d[2].item() / G
            h_norm = d[3].item() / G
            x_c = (gx + x_rel) * cell_size
            y_c = (gy + y_rel) * cell_size
            w, h = w_norm * img_size, h_norm * img_size
            x1, y1 = x_c - w / 2, y_c - h / 2
            cls = int(torch.argmax(pred[gy, gx, 1 + 4 * R:]).item())
            boxes.append((x1, y1, x1 + w, y1 + h, c, cls))
    return boxes

def decode_gt_boxes(target, img_size, grid_size):
    """Decode ground-truth target tensor to pixel-space boxes."""
    cell_size = img_size / grid_size
    boxes = []
    for gy in range(grid_size):
        for gx in range(grid_size):
            if target[gy, gx, 0].item() != 1.0:
                continue
            x_rel, y_rel, w_norm, h_norm = target[gy, gx, 1:5].tolist()
            x_c = (gx + x_rel) * cell_size
            y_c = (gy + y_rel) * cell_size
            w, h = w_norm * img_size, h_norm * img_size
            x1, y1 = x_c - w / 2, y_c - h / 2
            cls = int(torch.argmax(target[gy, gx, 5:]).item())
            boxes.append((x1, y1, x1 + w, y1 + h, cls))
    return boxes

def visualize_samples(model, dataloader, device, num_samples=3, conf_thresh=0.5):
    model.eval()
    images, targets = next(iter(dataloader))
    images  = images[:num_samples]
    targets = targets[:num_samples]
    with torch.no_grad():
        preds_out = model(images.to(device)).cpu()

    fig, axes = plt.subplots(num_samples, 2, figsize=(12, 4 * num_samples))
    if num_samples == 1:
        axes = [axes]

    for i in range(num_samples):
        img_np = images[i].permute(1, 2, 0).numpy()
        sz = img_np.shape[1]

        for col, (title, is_gt) in enumerate([("Ground Truth", True), ("Predictions", False)]):
            ax = axes[i][col]
            ax.imshow(img_np)
            ax.set_title(f"Sample {i+1} — {title}")
            ax.axis('off')

            if is_gt:
                for (x1, y1, x2, y2, cls) in decode_gt_boxes(targets[i], sz, GRID_SIZE):
                    ax.add_patch(patches.Rectangle(
                        (x1, y1), x2-x1, y2-y1, lw=2, edgecolor='lime', facecolor='none'))
                    ax.text(x1, y1 - 3, _cls_label(cls), color='lime', fontsize=7,
                            bbox=dict(facecolor='black', alpha=0.4, pad=1, edgecolor='none'))
            else:
                for (x1, y1, x2, y2, conf, cls) in decode_boxes_v2(
                        preds_out[i], sz, GRID_SIZE, REG_MAX, conf_thresh):
                    ax.add_patch(patches.Rectangle(
                        (x1, y1), x2-x1, y2-y1, lw=2, edgecolor='red', facecolor='none'))
                    ax.text(x1, y1 - 3, f"{_cls_label(cls)} {conf:.2f}", color='red', fontsize=7,
                            bbox=dict(facecolor='black', alpha=0.4, pad=1, edgecolor='none'))

    plt.tight_layout()
    plt.show()

visualize_samples(model, dataloader, device, conf_thresh=0.5)
```
